```python
from pathlib import Path
import json
import pandas as pd

OUT = Path('research/output/gear22_latency_edge_survival')
print((OUT / 'verdict.txt').read_text())
print(json.dumps(json.loads((OUT / 'power_gate.json').read_text()), indent=2))
```

```python
res = json.loads((OUT / 'episode_results.json').read_text())
rows = []
for r in res:
    if r.get('status') != 'ok':
        continue
    rows.append({
        'episode_id': r['episode_id'],
        'coin': r['base_coin'],
        'direction': r['direction'],
        'z0': r['z0'],
        'x0': r['x0'],
        'E_signal': r['E_signal'],
        'bin': r['E_signal_bin'],
        'contam': r['contaminated'],
        'econ': r['economic_primary'],
    })
pd.DataFrame(rows).sort_values('E_signal', ascending=False)
```

```python
print('state_at_L')
print(json.dumps(json.loads((OUT / 'summary_state_at_L.json').read_text()), indent=2))
print('bootstrap p_survive CI')
bs = json.loads((OUT / 'bootstrap_state.json').read_text())
for L, v in bs.get('L', {}).items():
    print(L, v.get('p_survive_ci'), 'point', v['point'].get('p_survive'), 'Rmed', v['point'].get('R', {}).get('median'))
pd.read_csv(OUT / 'per_episode_primary.csv')
```
