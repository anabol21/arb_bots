```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('/Users/mishatrubik/Desktop/spread/research/gear23_funding/data/market_funding/by_coin.csv')
data = data.dropna()
data
```

```python
# Гистограммы распределения средних значений по всем монетам
columns = ['rate_diff_pct', 'bybit_rate_pct', 'okx_rate_pct']

mean_by_coin = data.groupby('base_coin')[columns].mean()

fig, axs = plt.subplots(1, 3, figsize=(18, 5))
titles = [f'Среднее {col} по монетам' for col in columns]

for i, col in enumerate(columns):
    axs[i].hist(mean_by_coin[col], bins=50, edgecolor='black', alpha=0.7)
    axs[i].set_title(titles[i])
    axs[i].set_xlabel(col)
    axs[i].set_ylabel('Count')

plt.tight_layout()
plt.show()
```

```python
# Выберите монету для анализа
fixed_coin = 'TRUMP'  # Замените на желаемую монету

coin_data = data[data['base_coin'] == fixed_coin]

fig, axs = plt.subplots(1, 3, figsize=(18, 5))
columns = ['rate_diff_pct', 'bybit_rate_pct', 'okx_rate_pct']
titles = [f'{col} ({fixed_coin})' for col in columns]

for i, col in enumerate(columns):
    axs[i].hist(coin_data[col], bins=100, edgecolor='black', alpha=0.7)
    axs[i].set_title(titles[i])
    axs[i].set_xlabel(col)
    axs[i].set_ylabel('Count')

plt.tight_layout()
plt.show()
```
